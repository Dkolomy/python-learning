```python
import pandas as pd
```

```python
gdp_bln = pd.Series(data={
    'USA':28781,
    'Russia': 2056,
    'China': 18532,
    'South Korea': 1768,
    'France': 3130,
    'Germany': 4591,
    'UK': 3495,
    'Japan': 4110,
    'Canada': 2242,
})
```

```python
gdp_as_pct = gdp_bln / gdp_bln.max() * 100
```

```python
rounded_series = gdp_as_pct.apply(lambda x: f"{round(x, 2)}%")
```

```python
def convert(usd_value: float,
            currency: str,
            exchange_rates: dict[str, float]) -> float:
    if currency not in exchange_rates.keys():
        raise ValueError(f"Invalid currency: {currency}")
    return usd_value * exchange_rates[currency]
```

```python
print('\nConvert using exchange rate:')
converted_gdp = gdp_bln.apply(func=convert, currency='EUR', exchange_rates={'GBR': 1.1, 'EUR': 1.23})
print(converted_gdp)
```
